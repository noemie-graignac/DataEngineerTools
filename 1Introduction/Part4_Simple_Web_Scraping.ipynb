# Requête HTTP 

Un requête HTTP est une requête basée sur le protocole TCP, elle fait partie de la couche application de la couche OSI. Elle permet d'accéder aux données mise à disposition sur une adresse IP (ou url résolue par un DNS) et un port. 

Les deux ports les plus utilisés dans le web sont le 80 pour les sites en HTTP et le 443 pour les sites en HTTPS. HTTPS est une variable du protocole HTTP basé sur le protocole TLS.

Il existe de nombreux types de requêtes selon la convention `REST`: 
- GET
- POST
- PUT 
- DELETE
- UPDATE.

Dans notre cas, nous allons utiliser la plupart du temps des GET et potentiellement des POST. 
- Le GET permet comme son nom l'indique de récupérer des informations en fonction de certains paramètres. 
- Le POST nécessite un envoi de données pour récupérer des données. Le body du post est, la plupart du temps, envoyé sous la forme d'un objet JSON.

Ces requêtes encapsulent un certain nombre de paramètres qui permettent soient d'identifier une provenance et un utilisateur ou de réaliser différentes actions.

````python
#exo 1
import requests

class HTTPClient:

    def __init__(self):
        self.user_agent = "Mozilla/5.0"

    def get(self, url, timeout=10, retries=3):

        headers = {
            "User-Agent": self.user_agent
        }

        try:
            response = requests.get(
                url,
                headers=headers,
                timeout=timeout
            )

            response.raise_for_status()

            return response

        except requests.exceptions.RequestException as e:

            if retries > 0:

                print("Erreur, nouvelle tentative...")

                return self.get(
                    url,
                    timeout=timeout,
                    retries=retries - 1
                )

            else:
                print("Nombre maximum de tentatives atteint.")
                raise
````

````python
# EXERCICE 3

import requests
import random
from bs4 import BeautifulSoup
from urllib.parse import urljoin, urlparse

class HTTPClient:

    def __init__(self):
        self.user_agents = [
            "Mozilla/5.0 (Windows NT 10.0; Win64; x64) Chrome/120.0",
            "Mozilla/5.0 (X11; Linux x86_64) Firefox/120.0",
            "Mozilla/5.0 (Macintosh; Intel Mac OS X 10_15_7) Safari/605.1.15"
        ]

    def get(self, url, timeout=10, retries=3):
        headers = {
            "User-Agent": random.choice(self.user_agents)
        }

        try:
            response = requests.get(url, headers=headers, timeout=timeout)
            response.raise_for_status()
            return response

        except requests.exceptions.RequestException:
            if retries > 0:
                print("Nouvelle tentative...")
                return self.get(url, timeout, retries - 1)
            else:
                raise

    def get_soup(self, url):
        response = self.get(url)
        return BeautifulSoup(response.text, "html.parser")


# TEST DE LA CLASSE

url = "https://www.esiee.fr/"

client = HTTPClient()
soup = client.get_soup(url)


# 1. Récupérer le titre de la page

titre = soup.title.get_text(strip=True) if soup.title else "Aucun titre"
print("TITRE :", titre)


# 2. Récupérer tous les H1

h1 = [titre.get_text(strip=True) for titre in soup.find_all("h1")]

print("\nH1 :", h1)


# 3. Récupérer les liens des images

images = []

for image in soup.find_all("img"):
    src = image.get("src")
    if src:
        images.append(urljoin(url, src))

print("\nIMAGES :", images)


# 4. Récupérer les liens externes

liens_externes = []

domaine_principal = urlparse(url).hostname

for lien in soup.find_all("a", href=True):
    href = urljoin(url, lien["href"])
    domaine = urlparse(href).hostname

    if (
        urlparse(href).scheme in ("http", "https")
        and domaine
        and domaine != domaine_principal
    ):
        liens_externes.append(href)

print("\nLIENS EXTERNES :", list(set(liens_externes)))


# 5. Récupérer le texte principal

soup_texte = BeautifulSoup(str(soup), "html.parser")

for element in soup_texte(["script", "style", "nav", "footer", "header"]):
    element.decompose()

contenu = soup_texte.find("main")

if contenu is None:
    contenu = soup_texte.body or soup_texte

texte_principal = " ".join(contenu.get_text(separator=" ").split())

print("\nTEXTE PRINCIPAL :")
print(texte_principal[:2000])
````

# Exploitation des appels d'API



Losque le front du site récupère des données sur une API gérée par le back, un appel d'API est réalisé. Cet appel est recensé dans les appels réseaux. Il est alors possible de re-jouer cet appel pour récupérer à nouveau les données. Il est très facile de récupérer ces appels dans l'onglet Network de la console développeur de Chrome ou FireFox. La console vous permet de copier le code CURL de la requête et vous pouvez ensuite la transformer en code Python depuis le site https://curl.trillworks.com/.

Souvent les APIs sont bloquées avec certains paramètres. L'API vérifie que dans les headers de la requête HTTP ces paramètres sont présents :
* un token généré à la volée avec des protocoles OAuth2 (ou moins développés).
* un referer provenant du site web (la source de la requête), très facile à falsifier.

- Utiliser les informations développées plus haut pour récupérer les premiers résultats d'une recherche d'une requête
sur Google. 

Tips : 

- Ouvrir les outils de développements de Chrome ou Firefox
- Onglet Network
- Fouiller dans les requêtes pour voir à quoi ressemble un appel API Google
- Utilisez beautiful soup pour convertir le contenu de la request en objet et accéder aux balises

````python
# EXERCICE 4 : Récupérer les résultats d'une recherche Google

import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin, urlparse

# Recherche Google
recherche = "Data Engineering Python"

url = "https://www.google.com/search"

headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept-Language": "fr-FR,fr;q=0.9"
}

params = {
    "q": recherche,
    "num": 10,
    "hl": "fr"
}

# Envoyer la requête
response = requests.get(
    url,
    headers=headers,
    params=params,
    timeout=10
)

response.raise_for_status()

# Convertir le HTML avec BeautifulSoup
soup = BeautifulSoup(response.text, "html.parser")

# Récupérer les résultats
resultats = []

for titre in soup.find_all("h3"):

    lien = titre.find_parent("a")

    if lien and lien.get("href"):

        adresse = urljoin(response.url, lien["href"])

        if urlparse(adresse).scheme in ("http", "https"):

            resultats.append({
                "titre": titre.get_text(strip=True),
                "url": adresse
            })

# Afficher les résultats
print("Recherche :", recherche)
print("Nombre de résultats trouvés :", len(resultats))

for i, resultat in enumerate(resultats[:10], start=1):
    print(f"\nRésultat {i}")
    print("Titre :", resultat["titre"])
    print("URL :", resultat["url"])

if not resultats:
    print("\nAucun résultat extrait.")
    print("Google peut demander JavaScript, un consentement ou un CAPTCHA.")
    print("URL obtenue :", response.url)
````

Exercice Final
Utilisez tout ce que vous avez appris pour récupérer des articles de News avec une catégorie. Il est souvent intéressant de partir des flux RSS pour commencer :

Les données doivent comprendre :
- Le texte important propre
- L'url
- Le domaine
- la catégorie
- Le titre de l'article
- Le titre de la page
- (Facultatif) : les images

Tips : 

- Taper le nom de votre média favoris + RSS (par exemple : https://www.lemonde.fr/rss/)
- Aller dans le DOM de la page 
- Trouver les catégories et les liens vers les articles

````python
# EXERCICE FINAL : Récupération d'articles de presse

import requests
from bs4 import BeautifulSoup
from urllib.parse import urlparse, urljoin
import pandas as pd
import re
import time

# 1. URL du flux RSS de France 24
rss_url = "https://www.france24.com/fr/rss"

headers = {
    "User-Agent": "DataEngineerStudent/1.0 (projet pédagogique)"
}

# 2. Fonction pour nettoyer le texte
def nettoyer_texte(texte):
    return re.sub(r"\s+", " ", texte).strip()


# 3. Fonction pour récupérer les informations d'un article
def recuperer_article(url, categorie):

    response = requests.get(
        url,
        headers=headers,
        timeout=15
    )

    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")

    # Titre de la page
    titre_page = (
        soup.title.get_text(strip=True)
        if soup.title else ""
    )

    # Titre de l'article
    h1 = soup.find("h1")

    titre_article = (
        h1.get_text(" ", strip=True)
        if h1 else ""
    )

    # Texte principal
    contenu = (
        soup.find("article")
        or soup.find("main")
    )

    if contenu:
        copie = BeautifulSoup(str(contenu), "html.parser")

        for element in copie([
            "script", "style", "nav",
            "footer", "header", "aside"
        ]):
            element.decompose()

        texte = nettoyer_texte(
            copie.get_text(" ", strip=True)
        )
    else:
        texte = ""

    # Domaine
    domaine = urlparse(url).hostname

    # Images
    images = []

    if contenu:
        for image in contenu.find_all("img"):
            src = image.get("src")

            if src:
                images.append(urljoin(url, src))

    return {
        "titre_article": titre_article,
        "titre_page": titre_page,
        "url": url,
        "domaine": domaine,
        "categorie": categorie,
        "texte": texte,
        "images": images
    }


# 4. Récupération du flux RSS

response = requests.get(
    rss_url,
    headers=headers,
    timeout=15
)

response.raise_for_status()

rss = BeautifulSoup(response.content, "xml")

items = rss.find_all("item")

print("Nombre d'articles dans le RSS :", len(items))


# 5. Récupération des articles

articles = []

for item in items[:10]:

    lien = item.find("link")

    if not lien:
        continue

    url_article = lien.get_text(strip=True)

    categorie_tag = item.find("category")

    categorie = (
        categorie_tag.get_text(strip=True)
        if categorie_tag
        else "Actualités"
    )

    try:
        article = recuperer_article(
            url_article,
            categorie
        )

        # Utiliser le titre du RSS si le H1 est absent
        if not article["titre_article"]:
            titre_rss = item.find("title")

            article["titre_article"] = (
                titre_rss.get_text(strip=True)
                if titre_rss else ""
            )

        articles.append(article)

        print("Article récupéré :", article["titre_article"])

    except requests.exceptions.RequestException as e:
        print("Erreur pour", url_article, ":", e)

    time.sleep(1)


# 6. Création d'un DataFrame

df = pd.DataFrame(
    articles,
    columns=[
        "titre_article",
        "titre_page",
        "url",
        "domaine",
        "categorie",
        "texte",
        "images"
    ]
)

# 7. Affichage des résultats

display(df)


# 8. Sauvegarde dans un fichier CSV

df.to_csv(
    "articles_news.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nNombre d'articles récupérés :", len(df))
print("Fichier enregistré : articles_news.csv")
````
